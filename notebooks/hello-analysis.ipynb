```python
# Happy path — deterministic, stdlib-only analysis (no third-party imports).
import json

# Mirror of the greet() trim contract, expressed in Python.
samples = ["  workflow-lab  ", "Cloud", "\tlab\t"]
greeted = [s.strip() for s in samples if s.strip()]
assert greeted == ["workflow-lab", "Cloud", "lab"], greeted

print("happy path OK:", greeted)
print("json round-trip OK:", json.loads(json.dumps(greeted)) == greeted)
```

```python
# Edge path — whitespace-only / non-string input must be rejected (error path).
def require_name(name):
    if not isinstance(name, str) or not name.strip():
        raise TypeError("name must be a non-empty string")
    return name.strip()

# 1) whitespace-only raises
try:
    require_name("   ")
except TypeError as exc:
    print("edge 1 OK:", exc)
else:
    raise AssertionError("expected TypeError for whitespace-only")

# 2) non-string raises
try:
    require_name(123)
except TypeError as exc:
    print("edge 2 OK:", exc)
else:
    raise AssertionError("expected TypeError for non-string")

# 3) happy still works
assert require_name("  lab  ") == "lab"
print("edge contract OK")
```
